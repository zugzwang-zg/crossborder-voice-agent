# 01｜Data cleaning check

使用公开的 10 条均衡样本检查字段、语言和星级分布。完整清洗数据不进入公开仓库。

```python
from collections import Counter
from pathlib import Path
import csv

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sample_path = PROJECT_ROOT / "data" / "sample" / "demo_reviews.csv"
with sample_path.open(encoding="utf-8-sig", newline="") as handle:
    rows = list(csv.DictReader(handle))

print("records:", len(rows))
print("languages:", Counter(row["language"] for row in rows))
print("stars:", Counter(int(row["stars"]) for row in rows))
print("fields:", list(rows[0]))
```

预期结果：10 条评论；英语与西班牙语各 5 条；1—5 星各 2 条。正式 1,000 条数据使用同一分层逻辑。
