# 03｜Evaluation

从评测报告 JSON 读取冻结 100 条金标测试集上的 Baseline v1 与 Improved v9 指标。

```python
from pathlib import Path
import json

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
report_path = PROJECT_ROOT / "reports" / "evaluation_report.json"
report = json.loads(report_path.read_text(encoding="utf-8"))
baseline = report["overall_metrics"]["baseline"]
improved = report["overall_metrics"]["improved"]
```

```python
rows = [
    ("JSON success", baseline["json_parse_success_rate"], improved["json_parse_success_rate"]),
    ("Sentiment accuracy", baseline["sentiment_accuracy"], improved["sentiment_accuracy"]),
    ("Sentiment Macro-F1", baseline["sentiment_macro_f1"], improved["sentiment_macro_f1"]),
    ("Aspect F1", baseline["aspect_f1"], improved["aspect_f1"]),
    ("Issue F1", baseline["issue_f1"], improved["issue_f1"]),
    ("Semantic evidence", baseline["manual_semantic_evidence_validity"], improved["manual_semantic_evidence_validity"]),
]
print(f"{'Metric':<24} {'Baseline':>10} {'Improved':>10}")
for name, before, after in rows:
    print(f"{name:<24} {before:>9.2%} {after:>9.2%}")
```

人工证据语义有效率来自每个版本各 25 个证据项的人工复核；商业建议评分来自每个版本各 20 条建议。
