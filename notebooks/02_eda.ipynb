# 02｜Exploratory analysis

读取不含完整评论原文的聚合表，复现低星痛点、高星购买动机和跨语言关注差异。

```python
from pathlib import Path
import csv

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

def read_csv(relative_path):
    with (PROJECT_ROOT / relative_path).open(encoding="utf-8-sig", newline="") as handle:
        return list(csv.DictReader(handle))

pain_points = read_csv("data/aggregated/low_star_pain_points.csv")
drivers = read_csv("data/aggregated/high_star_purchase_drivers.csv")
language_gaps = read_csv("data/aggregated/language_focus_comparison.csv")
```

```python
print("Top low-star pain points")
for row in pain_points[:5]:
    print(f"- {row['name']}: {row['support_reviews']} reviews")

print("\nTop high-star purchase motivations")
for row in drivers[:5]:
    print(f"- {row['name']}: {row['support_reviews']} reviews")

print("\nLargest language attention gaps")
for row in language_gaps[:5]:
    gap = float(row['absolute_rate_gap']) * 100
    print(f"- {row['aspect_name']}: {gap:.1f} pp, higher in {row['higher_attention_language']}")
```

这些是描述性结果。样本内差异不能直接解释为国家、文化或全部消费者群体之间的因果差异。
